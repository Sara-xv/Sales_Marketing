```python
# Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Import sales marketing dataset for analysis
data = pd.read_csv('sales_marketing_analysis1.csv')
```

```python
data.head()
```

```python
# Display dataset information
data.info()
```

```python
# Convert columns to categorical type for memory efficiency and better performance
data['Customer_Segment'] = data['Customer_Segment'].astype('category')
data['Month'] = data['Month'].astype('category')
```

```python
# Check for missing values
data.isnull().sum()
```

**✅ Missing Values Check:** No missing values detected in the dataset. This step can be skipped.

```python
# DATA QUALITY CONTROL & VALIDATION WITH REMOVED RECORDS TRACKING
 
# Create a copy of original data before filtering
original_data = data.copy()
initial_count = len(original_data)

# Apply multiple filters to remove invalid records
filter_mask = (
    # ✅ Sales Amount Validation
    (data['Sales_Amount'] > 0) &
    
    # ✅ Quantity Validation
    (data['Quantity'] >= 1) &
    
    # ✅ Discount Percent Validation
    (data['Discount_Percent'] >= 0) & (data['Discount_Percent'] <= 100) &
    
    # ✅ Customer Age Validation
    (data['Customer_Age'] >= 10) & (data['Customer_Age'] <= 100) &
    
    # ✅ Conversion Rate Validation
    (data['Conversion_Rate'] >= 0) & (data['Conversion_Rate'] <= 100) &
    
    # ✅ Customer Satisfaction Score Validation
    (data['Customer_Satisfaction_Score'] >= 1) & 
    (data['Customer_Satisfaction_Score'] <= 5) &
    
    # ✅ Additional Validation for missing values
    (data['Sales_Amount'].notna()) &
    (data['Quantity'].notna()) &
    (data['Customer_Age'].notna())
)

# Get removed records (inverse of the filter mask)
removed_data = data[~filter_mask].copy()

# Apply the filter to keep only valid records
data = data[filter_mask]

# 🔍 Analyze removed records
final_count = len(data)
removed_count = len(removed_data)
removed_percentage = (removed_count / initial_count) * 100

print("=" * 60)
print("DATA QUALITY REPORT")
print("=" * 60)
print(f"Initial record count: {initial_count:,}")
print(f"Records after filtering: {final_count:,}")
print(f"Records removed: {removed_count:,} ({removed_percentage:.2f}%)")
print("=" * 60)

# 📊 Detailed analysis of removed records by reason
if not removed_data.empty:
    print("\nDETAILED ANALYSIS OF REMOVED RECORDS:")
    print("-" * 40)
    
    # Check which validation rules each removed record violated
    reasons = {
        'Sales_Amount ≤ 0': (removed_data['Sales_Amount'] <= 0) | (removed_data['Sales_Amount'].isna()),
        'Quantity ≤ 1': (removed_data['Quantity'] <= 1) | (removed_data['Quantity'].isna()),
        'Discount_Percent out of range': (removed_data['Discount_Percent'] < 0) | 
                                        (removed_data['Discount_Percent'] > 100) |
                                        (removed_data['Discount_Percent'].isna()),
        'Customer_Age out of range': (removed_data['Customer_Age'] < 10) | 
                                    (removed_data['Customer_Age'] > 100) |
                                    (removed_data['Customer_Age'].isna()),
        'Conversion_Rate out of range': (removed_data['Conversion_Rate'] < 0) | 
                                       (removed_data['Conversion_Rate'] > 100) |
                                       (removed_data['Conversion_Rate'].isna()),
        'Satisfaction_Score out of range': (removed_data['Customer_Satisfaction_Score'] < 1) | 
                                          (removed_data['Customer_Satisfaction_Score'] > 5) |
                                          (removed_data['Customer_Satisfaction_Score'].isna())
    }
    
    # Count violations by reason
    for reason, mask in reasons.items():
        count = mask.sum()
        if count > 0:
            print(f"{reason}: {count:,} records ({count/removed_count:.1%})")
    
    # Show records with multiple violations
    multi_violation_mask = sum(reasons.values()) > 1
    multi_violation_count = multi_violation_mask.sum()
    if multi_violation_count > 0:
        print(f"\nRecords with multiple violations: {multi_violation_count:,}")
        print("\nSample of records with multiple violations:")
        print(removed_data[multi_violation_mask].head(3).to_string())
    
    # 💡 Save removed records to CSV for further investigation
    removed_data.to_csv('removed_records_analysis.csv', index=False)
    print(f"\n📁 Removed records saved to 'removed_records_analysis.csv' for detailed investigation")

# 💡 Warning if significant data loss
if removed_percentage > 5:
    print(f"\n⚠️  WARNING: {removed_percentage:.1f}% of data removed!")
    print("   Consider investigating data quality issues.")
else:
    print(f"\n✅ Data quality acceptable ({removed_percentage:.1f}% removed)")

print("=" * 60)
```

```python
#  Create age group categories for customer segmentation
conditions = [
    (data['Customer_Age'] >= 18) & (data['Customer_Age'] <= 20),
    (data['Customer_Age'] >= 21) & (data['Customer_Age'] <= 30),
    (data['Customer_Age'] >= 31) & (data['Customer_Age'] <= 45),
    (data['Customer_Age'] >= 46) & (data['Customer_Age'] <= 60),
    (data['Customer_Age'] >= 61)
]

choices = ['18-20','21-30','31-45','46-60','61+']

data['Age_Group'] = np.select(conditions, choices, default='Unknown')
```

```python
# ROL =  Return on Investment
def calculate_profit(x):
    return x['Total_Sales'] - x['Total_Ad_Cost']

 
ROL = data.groupby('Campaign_Channel')[['Ad_Cost', 'Sales_Amount']].sum().reset_index()
ROL = ROL.rename(columns={
    'Campaign_Channel': 'Channel',
    'Ad_Cost': 'Total_Ad_Cost',
    'Sales_Amount': 'Total_Sales'
})

ROL['Total_Profit'] = ROL.apply(calculate_profit, axis=1)
ROL
```
